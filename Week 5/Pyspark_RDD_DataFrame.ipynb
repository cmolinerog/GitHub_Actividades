```python
# Q1. Create a SparkSession named RDD_DataFrame_Lab.
import os

os.environ["SPARK_LOCAL_IP"] = "127.0.0.1"
os.environ["PYSPARK_PYTHON"] = "python"
os.environ["PYSPARK_DRIVER_PYTHON"] = "python"

from pyspark.sql import SparkSession

spark = SparkSession.builder \
    .appName("RDD_DataFrame_Lab") \
    .master("local[1]") \
    .getOrCreate()
```

```python
# Q2. Load the customers.csv file into a DataFrame using Header=True and InferSchema=True.
df = spark.read.csv("C:\pyspark\customers.csv", header=True,
    inferSchema=True)
```

```python
#Q3. Display the first 10 records and print the schema of the DataFrame.
df.show(10)
df.printSchema()
```

```python
#Q4. Find the total number of records present in the dataset.
df.count()
```

```python
df = df.withColumnRenamed("d", "CUSTOMERID")
df.show()
```

```python
#Q5. Display only the columns: CustomerID, CustomerName, Country.
df.select("CUSTOMERID", "CUSTOMERNAME", "COUNTRY").show()
```

```python
# Q6. Filter all customers whose age is greater than 30
from pyspark.sql.functions import col

df_30 = df.filter(col("CUSTOMERID") > 30).select("CUSTOMERNAME", "CUSTOMERID")
df_30.show()
```

```python
# Q7. Find all customers belonging to India.
df_india = df.filter(col("COUNTRY") == "INDIA").select("CUSTOMERNAME", "COUNTRY")
df_india.show()
```

```python
# Q8. Count the number of customers from each country.
df_country = df.groupBy("COUNTRY").count()
df_country.show()
```

```python
# Q9. Sort customers by Age in descending order.
from pyspark.sql.functions import floor, rand
df = df.withColumn("AGE", floor(rand() * 60 + 18)) #rand genera números aleatorios, floor convierte a entero, 60 + 18 es el rango
df_age = df.sort("AGE", ascending = False)
df_age.show()
```

```python
# Q10. Find the average age of customers.
from pyspark.sql.functions import avg

df.select(avg("AGE").alias("average_age")).show()
```

```python
# Q11. Convert the DataFrame into an RDD.
rdd = df.rdd
rdd.collect()
```

```python
# Q12. Print the type of the converted object.

print(type(rdd))
```

```python
# Q13. Display the first 5 records from the RDD.
print(rdd.take(5))
```

```python
# Q14. Using RDD operations, extract CustomerName and Email.
rdd1 = rdd.map(lambda row:(row["CUSTOMERNAME"], row["EMAIL"]))
rdd1.collect()
```

```python
# Q15. Create an RDD containing only customers whose age is above 35.
rdd_customer_35 = rdd \
    .filter(lambda row:row["AGE"] > 35) \
    .map(lambda x:(x["CUSTOMERNAME"], x["AGE"]))
rdd_customer_35.collect()
```

```python
# Q16. Using RDD transformations, extract email domains from customer emails.
rdd_email = rdd.map(lambda x:x["EMAIL"].split("@")[1])
rdd_email.collect()
```

```python
# Q17. Count how many customers belong to each email domain using RDD operations.
rdd_email_count = rdd.map(lambda x:x["EMAIL"].split("@")[1]) \
                .map(lambda w:(w,1)) \
                .reduceByKey(lambda x,y:(x+y))

rdd_email_count.collect()
```

```python
# Q18. Create an RDD containing (CustomerID, Country).
rdd_country = rdd.map(lambda x:(x["CUSTOMERID"],x["COUNTRY"]))
rdd_country.collect()
```

```python
# Q19. Use an action to display the first 10 records from the RDD.
rdd_country.take(10)
```

```python
# Q22. Convert the filtered age RDD back into a DataFrame.

df_new = spark.createDataFrame(rdd_customer_35, schema= ["CUSTOMERNAME", "AGE"]) #col name
df_new.show()
```

```python
# Q24. Show the top 5 records from the converted DataFrame.
df_new.show(5)
```

```python
# Q25. Create an RDD that extracts (CustomerID, EmailDomain).
rdd_email_domain = rdd.map(lambda x:(x["CUSTOMERID"], x["EMAIL"].split("@")[1]))
rdd_email_domain.collect()

```

```python
# Q26. Convert the above RDD into a DataFrame with columns CustomerID and EmailDomain.

df25 = spark.createDataFrame(rdd_email_domain, schema=["CUSTOMERID", "EMAIL_DOMAIN"])
df25.show()
```

```python
# Q27. Join the original DataFrame with the EmailDomain DataFrame using CustomerID.
df27 = df.join(df25, on="CUSTOMERID")
df27.show()

```

```python
# Q28. Display CustomerName, Email, and EmailDomain.
df28 = df27.select("CUSTOMERNAME", "EMAIL")
df28.show()
```

```python
# Q29. Use the explain() method on the final joined DataFrame.
df28.explain()
```

```python
# Q30. Identify the Logical Plan and Physical Plan from the explain output.
# Logical plan: what operations spark perform
    # Project → selecting columns (CUSTOMERNAME, EMAIL)
    # Filter isnotnull → removing null values
    # Join → joining datasets on CUSTOMERID
    # FileScan / Scan → reading data sources (CSV + RDD)
#Phisical Plan: how it is executed
    # AdaptiveSparkPlan → Spark is using Adaptive Query Execution (AQE) to optimize runtime
    # BroadcastHashJoin → Spark uses a broadcast join (efficient for small tables)
    # BroadcastExchange → sends small dataset to all executors
    # FileScan → reads CSV file (customers data)
    # Scan ExistingRDD → reads RDD-based DataFrame (email domains)
```

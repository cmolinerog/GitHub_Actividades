```python
df = spark.table("spotify_data")
df.show()
```

```python
# 1. Which is the longest song released yet? 
df_longest_song = df.sort("duration_ms", ascending = False)
df_longest_song.show()
```

```python
# 2. How many songs has Drake sung? 
df_drake = df.filter(df.artist == "Drake")
df_drake.show()
```

```python
# 3. List those songs with letter ‘Love’ in them. 
df_love = df.filter(df.song_title.contains("Love")).select("song_title")
df_love.show()
```

```python
# 4. Select songs which have a number in them.
df_song_number = df.filter(df.song_title.rlike(r"\d")) # \d significa “cualquier dígito del 0 al 9
df_song_number.show()
```

```python
# 5. You have a bachelorette weekend, which songs would you play the most?
df_bachelorette = df.filter((df.danceability > 0.7) & (df.liveness > 0.5)).sort("energy", ascending = False)
df_bachelorette.show(10)
```

```python
# 6. In a dance battle, which 3 songs would you choose for a dance off ?
df_dance = df.sort("danceability", ascending = False)
df_dance.show(3)
```

```python
# 7. Which artist has the max songs released?
df_max_artist = df.groupBy("artist").count().sort("count", ascending = False)
df_max_artist.show(1)
```

```python
#  8. Make a ‘break-up’ song list >..< 
df_breakup = df.filter((df.tempo < 70) & (df.energy < 0.3)).sort("loudness", ascending = True)
df_breakup.show(10)
```
